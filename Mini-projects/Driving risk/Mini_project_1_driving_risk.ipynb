## `Project description` <br>

### **Introduction**

Risk can be measured for car drivers via different indicators on roads. Traffic crashes have been used as the most common indicators of risk. However, crashes are rare events and should occur first so that their data can be collected and analysed. Therefore, relying on crashes is reactive and there is a need for proactive indicators of risk such as near-misses which are interactions at a step before crashes actually occur. Near-misses can be defined based on driving kinematics such as headway, speed, acceleration and so forth. Meanwhile, many factors are predictors of risk and contribute to (influence) risk. These factors arise from distinct sources such as drivers’ demographics, past behaviours, driving style, as well as road environment and traffic characteristics.

*The objective of this mini-project is to predict risky driving behaviour and identify factors that contribute the most to it.*<br>
