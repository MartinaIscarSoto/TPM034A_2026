## `Project description` <br>

### **Introduction**

Gait impairment is an important motor symptom of Parkinson’s disease, and its severity varies between patients. Shorter steps, reduced arm swing, more variable stride timing, a stooped posture and difficulties with turning can reflect gait impairment. Clinicians assess gait severity using the gait item of the Unified Parkinson’s Disease Rating Scale (UPDRS Part III). 3D motion capture records walking patterns precisely. Marker-based systems, (depth) cameras and pose estimation models or wearable inertial sensors track how the joints and body segments move during standardized walking trials. From these recordings, spatiotemporal measures such as stride length, cadence and double-support time, joint angle trajectories, and measures of symmetry and variability can be extracted. Machine learning models can use these movement characteristics to predict the clinician-rated gait severity score. The challenge is that clinics differ in their use of different recording systems, patient cohorts, and walking protocols. Their populations differ in age, disease stage and medication status.

*The objective of this mini-project is to predict gait severity (the clinician-rated UPDRS Part III gait score) from walking data from multiple clinics and identify the movement features that contribute most to the prediction.*<br>
