## `Project description` <br>

### **Introduction**

Populytics, a recently founded company, studies the preferences of Dutch citizens about public policy dilemmas (such as wind turbines, refugee centers, and health care budgeting) in the Netherlands by carrying out large-scale online consultations using the scientific PVE-method. By gaining insight into the worries and preferences of citizens, the company provides the government (and other decision makers) with helpful insights to make better decisions. Within these consultations, there are distinct groups of respondents: some with very clear opinion and some with moderate preferences. The former may shape the future policies whereas the latter may claim they haven’t actively voiced their opinion on the topic. There could be many reasons behind these group differences. Some respondents may be disinterested in a topic; some may be unknowledgeable; some may be well-considered and nuanced position; some may be reluctant to change, or realistic about social change. Or something else can be entirely at play.

*The objective of this mini-project is to predict groups of respondents and the underlying reasons behind such groups. Are there recognizable, coherent and recurrent subgroups? What questions or (demographic) factors allow you to identify these subgroups?*<br>
